## Data Structure :

```python
import numpy as np
# data of restaurant sales :

data = np.array([
    [1, 1200,1400,4322,4500,4599],#nanskar
    [2, 1144,4585,5265,4852,4585],#pappe
    [3, 7852,4586,9625,4852,3571],#cholewale
    [4, 7864,8569,4500,8035,2345],#dosaram
    [5, 7896,8522,7931,8016,1495],#hub
])

print("shape of array", data.shape)
print("Required Value:\n", data[0:3])
#this means all the row and coloumn starting from index 1 till end
print("Required Value in fancy way:\n", data[:,1:])


```

```python
# <-------------numpy logic------------------->
sales_per_year= np.sum(data[:,1:],axis=0)
print("Sales per year:\n", sales_per_year)


sales_per_year= np.sum(data,axis=1)
print("Sales Of Each restaurant:\n", sales_per_year)

# <------------python logic------------------->
# total sales 
from functools import reduce 
sales_nanskar= data[1,1:]
total_sales= reduce(lambda x,y:x+y,sales_nanskar)
print("total_sales nanskar " , total_sales)
```

## Minimum sales Per Restaurant

```python
# <---------Numpy Logic------------------->

data = np.array([
    [1, 1200,1400,4322,4500,4599],#nanskar
    [2, 1144,4585,5265,4852,4585],#pappe
    [3, 7852,4586,9625,4852,3571],#cholewale
    [4, 7864,8569,4500,8035,2345],#dosaram
    [5, 7896,8522,7931,8016,1495],#hub
])

min_sales_restaurant = np.min(data[:,1:], axis =1)
print(min_sales_restaurant)
#this gives minimum sales per restaurant
# <================same function for max as well ================>>

# <---------Python Logic------------------->
from functools import reduce
min_sales_per_year = data[:, 1]
filtered_sales = reduce(lambda x,y: y if x>y else x, min_sales_per_year)
print("Minimum Sales :",filtered_sales)
```

## Average and Max sales

```python
import numpy as np
from functools import reduce

average_sales_restaurant = np.mean(data[:, 1:], axis=0)
print(average_sales_restaurant)

# reduce can be used as well

# we use max instead of mean and the same function

# cummulative sum
cummulative_sum = np.cumsum(data[:, 1:], axis=1)
print("cummulative Sum :", cummulative_sum)
```

## Addition multiplication and dot product in numpy 

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector Addition :", vector1 + vector2)

print("Vector Multiplication :", vector1 * vector2)

print("Vector Dot Product :", np.dot(vector1,vector2)) 

angle = np.arccos(np.dot(vector1 , vector2)/ (np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print(angle)
```
